```python
from pathlib import Path

# Create all assignment files in the current folder
project = Path.cwd()

# 1. SPEECH-TO-TEXT
speech_code = '''
from faster_whisper import WhisperModel

def transcribe_audio(audio_path="audio.wav"):
    model = WhisperModel("tiny.en", device="cpu", compute_type="int8")
    segments, info = model.transcribe(audio_path, language="en")

    text = " ".join(segment.text.strip() for segment in segments)

    if not text:
        raise ValueError("No speech detected.")

    return text

if __name__ == "__main__":
    result = transcribe_audio()
    print("Transcribed Text:", result)
'''

# 2. SENTIMENT ANALYSIS
nlp_code = '''
from transformers import pipeline

def analyze_sentiment(text):
    model = pipeline(
        "sentiment-analysis",
        model="cardiffnlp/twitter-roberta-base-sentiment-latest"
    )

    result = model(text[:512])[0]

    return {
        "sentiment": result["label"].lower(),
        "score": round(float(result["score"]), 4)
    }

if __name__ == "__main__":
    text = "The weather is beautiful and I feel happy."
    print(analyze_sentiment(text))
'''

# 3. IMAGE LABELLING
image_code = '''
from transformers import pipeline

def label_image(image_path="photo.jpg"):
    model = pipeline(
        "image-classification",
        model="google/vit-base-patch16-224"
    )

    results = model(image_path, top_k=5)

    labels = []

    for item in results:
        labels.append({
            "label": item["label"],
            "confidence": round(float(item["score"]), 4)
        })

    return labels

if __name__ == "__main__":
    results = label_image()

    print("Top 5 Image Labels:")

    for item in results:
        print(item)
'''

# 4. COMBINED PIPELINE
pipeline_code = '''
import json
from datetime import datetime, timezone

from speech import transcribe_audio
from nlp import analyze_sentiment
from image import label_image

def save_log(service, input_data, output_data):

    record = {
        "timestamp": datetime.now(timezone.utc).isoformat(),
        "service": service,
        "input": input_data,
        "output": output_data
    }

    with open("api_calls.log", "a", encoding="utf-8") as file:
        file.write(json.dumps(record) + "\\\\n")


def run_pipeline():

    print("\\\\n--- SPEECH TO TEXT ---")

    text = transcribe_audio("audio.wav")
    print("Transcription:", text)

    save_log("speech", "audio.wav", text)

    print("\\\\n--- SENTIMENT ANALYSIS ---")

    sentiment = analyze_sentiment(text)
    print("Sentiment:", sentiment)

    save_log("nlp", text, sentiment)

    print("\\\\n--- IMAGE LABELLING ---")

    labels = label_image("photo.jpg")
    print("Image Labels:", labels)

    save_log("image", "photo.jpg", {"labels": labels})

    print("\\\\nAll tasks completed successfully!")
    print("Results saved in api_calls.log")


if __name__ == "__main__":
    run_pipeline()
'''

# 5. REQUIREMENTS
requirements = '''
faster-whisper
transformers
torch
pillow
'''

# 6. README
readme = '''
# Cloud AI Assignment

## Objective
To implement speech-to-text, sentiment analysis and image labelling
using pre-trained AI models on a local computer.

## Technologies
- Python
- Faster-Whisper
- Hugging Face Transformers
- PyTorch
- JSON Logging

## Files
- speech.py: Speech-to-text
- nlp.py: Sentiment analysis
- image.py: Image labelling
- pipeline.py: Combined execution
- api_calls.log: JSON-lines API-call log

## Execution
Run individual scripts or execute:

python pipeline.py

## Note
This project uses locally executed pre-trained models.
It does not require paid cloud API access.
'''

# Create the files
files = {
    "speech.py": speech_code,
    "nlp.py": nlp_code,
    "image.py": image_code,
    "pipeline.py": pipeline_code,
    "requirements.txt": requirements,
    "README.md": readme,
    "api_calls.log": ""
}

for filename, content in files.items():
    (project / filename).write_text(content.strip() + "\n", encoding="utf-8")

print("All seven project files created successfully!")
print("\nProject location:", project)
print("\nFiles created:")

for filename in files:
    print("✔", filename)
```
